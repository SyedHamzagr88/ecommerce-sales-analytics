```python
import sqlite3
import pandas as pd
import numpy as np 
```

```python
conn = sqlite3.connect("../database/ecommerce.db")
customers= pd.read_sql_query("SELECT * FROM customers",conn)
products = pd.read_sql_query(
    "SELECT * FROM Products",
    conn
)
orders = pd.read_csv("../data/Orders.csv")
payments = pd.read_csv("../data/Payments.csv")
```

```python
merged = orders.merge(products,on= "product_id")
```

```python
merged.head()
```

```python
#merged revenue feature
merged["revenue"]= (merged["quantity"]*merged["price"])
```

```python
merged[["product_name","quantity","price","revenue"]].head()
```

```python
merged["order_date"]=pd.to_datetime(merged["order_date"])
```

```python
merged["month"]=(merged["order_date"].dt.month)
```

```python
merged["month_name"]=(merged["order_date"].dt.month_name())
```

```python
merged["day_name"]=(merged["order_date"].dt.day_name())
```

```python
merged["year"]=(merged["order_date"].dt.year)
```

```python
merged["is_weekend"]=(merged["order_date"].dt.dayofweek>=5)
```
